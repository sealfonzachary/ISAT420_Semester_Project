# Problem Introduction
"Ghost Forests" are expanding in the Chesapeake Bay as coastal forests are exposed to increasing saltwater intrusion. This process is affecting the health of the Bay, as the death of salt insensitive trees of the bay decreases overall biodiversity, wildlife habitat, and coastal ecosystem stability. To better understand this problem, long term environmental data is needed to examine how salinity patterns relate to ghost forest encroachment over time.

## Problem Statement
The purpose of this project is to explore how environmental drivers create ghost forests in the Chesapeake Bay. Ghost forests are caused by two main drivers, salinity concentrations and relative sea level rise (RSLR). We expect our project to result in a refined understanding of how salinity and RSLR are associated with ghost forest expansion, identifying areas that are most impacted. This information can help infer where future ghost forests are likely to appear and can be used to better inform future land management decisions.


## Background
"Sea level rise-induced saltwater intrusion fundamentally alters coastal ecosystem functions by shifting plant communities from freshwater-adapted to salt-tolerant species, reducing biodiversity and changing nutrient cycling patterns. This salinization process disrupts established food webs, degrades freshwater wetland habitats that serve as critical nurseries for many species, and can lead to the formation of “ghost forests” where salt-intolerant trees die off, ultimately transforming productive coastal ecosystems into less diverse salt marshes or open water areas (Kirwan and Gedan, 2019)..." (Ruess et al., 2025).

## Research Question(s)
**Are areas with higher salinity and RSLR more likely to be associated with ghost forest expansion?**

This question seeks to answer how two main drivers, relative sea level rise and salinity concentrations, can change an ecosystem over time. Answering this question can help us infer where future ghost forests are likely to appear, which can have big implications for regional ecosystems
