```python
# --- 1. SETUP & CONFIGURATION ---
import requests
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import linregress
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# List of NOAA station IDs for the Chesapeake Bay area
station_ids = ["8635750", "8639348", "8638610", "8638901", "8632200", "8637689", "8571421",
               "8636580", "8577330", "8635027", "8571892", "8594900", "8575512", "8573364", 
               "8574680", "8573927"] 

# Dictionary to store results for the final map and summary table
station_data_summary = {}
```

```python
# --- 2. DATA PROCESSING LOOP ---
# Initialize the figure for time-series subplots
fig_ts, axes = plt.subplots(nrows=len(station_ids), ncols=1, 
                            figsize=(12, 4 * len(station_ids)), sharex=True)

for i, station_id in enumerate(station_ids):
    try:
        # A. Fetch Metadata (Station Name, Latitude, Longitude)
        meta_url = f"https://api.tidesandcurrents.noaa.gov/mdapi/prod/webapi/stations/{station_id}.json"
        meta_res = requests.get(meta_url)
        meta_res.raise_for_status()
        meta_json = meta_res.json()
        
        s_info = meta_json['stations'][0]
        s_name = f"{s_info['name']}, {s_info['state']}"
        s_lat, s_lon = float(s_info['lat']), float(s_info['lng'])
        
        print(f"Processing: {s_name}...")

        # B. Fetch Monthly Mean Sea Level Data
        params = {
            "begin_date": "19940101",
            "end_date": "20241231",
            "station": station_id,
            "product": "monthly_mean",
            "datum": "MSL",
            "units": "metric",
            "time_zone": "lst_ldt",
            "format": "json"
        }
        
        data_res = requests.get("https://api.tidesandcurrents.noaa.gov/api/prod/datagetter", params=params)
        data = data_res.json()

        if 'data' not in data:
            print(f"  ! No data found for {s_name}.")
            continue

        # C. Clean Data with Pandas
        df = pd.DataFrame(data['data'])
        df['timestamp'] = pd.to_datetime(df['year'] + "-" + df['month'])
        df['sea_level'] = pd.to_numeric(df['MSL'], errors='coerce')
        df.set_index('timestamp', inplace=True)
        df = df[['sea_level']].dropna()

        # D. Linear Regression (Calculate mm/yr rise)
        x = df.index.map(pd.Timestamp.toordinal)
        y = df['sea_level']
        slope, intercept, r_val, p_val, std_err = linregress(x, y)
        rate_mm_yr = slope * 365.25 * 1000 # Convert daily meters to yearly mm

        # Store results for the map
        station_data_summary[station_id] = {
            "name": s_name, "lat": s_lat, "lon": s_lon, "rate": rate_mm_yr
        }

        # E. Plot Individual Station Trend
        ax = axes[i]
        ax.plot(df.index, df['sea_level'], label='Monthly Mean', alpha=0.3, color='gray')
        ax.plot(df.index, df['sea_level'].rolling(window=12).mean(), label='12-Mo Avg', color='red')
        ax.plot(df.index, intercept + slope * x, color='green', linewidth=2, 
                label=f'Trend: {rate_mm_yr:.2f} mm/yr')
        ax.set_title(f"Station: {s_name}")
        ax.set_ylabel("Meters (MSL)")
        ax.legend(loc='upper left')

    except Exception as e:
        print(f"Error processing {station_id}: {e}")

# Save the time-series plots
plt.tight_layout()
```

```python
# --- 3. SUMMARY TABLE ---
# 1. Convert the dictionary to a Pandas DataFrame
# We use orient='index' because the station IDs are the keys of your dictionary
RSLR_df = pd.DataFrame.from_dict(station_data_summary, orient='index')

# 2. Reset the index so 'Station ID' becomes a normal column instead of the row label
RSLR_df.index.name = 'Station ID'
RSLR_df = RSLR_df.reset_index()

# 3. Rename columns to match your desired format
RSLR_df.columns = ['Station ID', 'Area', 'Lat', 'Lon', 'Rate (mm/yr)']

# 4. Export the DataFrame to CSV
RSLR_df.to_csv('RSLR_Trend_Summary.csv', index=False)
print("\nDetailed summary saved to 'RSLR_Trend_Summary.csv'")

RSLR_df
```

```python
# --- 4. GEOGRAPHIC MAP (CARTOPY) ---
lons = [info['lon'] for info in station_data_summary.values()]
lats = [info['lat'] for info in station_data_summary.values()]
rates = [info['rate'] for info in station_data_summary.values()]

fig_map = plt.figure(figsize=(8, 10))
ax_map = plt.axes(projection=ccrs.PlateCarree())

# Add map features
ax_map.add_feature(cfeature.COASTLINE)
ax_map.add_feature(cfeature.STATES, edgecolor='gray')
ax_map.add_feature(cfeature.RIVERS)
ax_map.set_extent([-77.5, -75.0, 36.5, 40.0], crs=ccrs.PlateCarree())

# Plot station points colored by RSLR rate
scatter = ax_map.scatter(lons, lats, c=rates, cmap='viridis', s=120, 
                         edgecolor='black', transform=ccrs.PlateCarree(), zorder=5)

# Add colorbar and title
plt.colorbar(scatter, orientation='vertical', pad=0.02, label='RSLR Rate (mm/year)')
plt.title("Relative Sea Level Rise Rates: Chesapeake Bay (1994-2024)")

plt.show()
```

```python
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from io import StringIO

# 1. Corrected REST API call
# Removed 'parameterCode' as it's not valid for the /site/ endpoint
usgs_url = "https://waterservices.usgs.gov/nwis/site/"
params = {
    'format': 'rdb',
    'bBox': '-77.5,36.5,-75.5,39.5',
    'siteStatus': 'active',
    'seriesCatalogOutput': 'true', # This adds date ranges to your dataframe
    'outputDataTypeCd': 'iv' 
}
response = requests.get(usgs_url, params=params)

# Check if the request was successful
if response.status_code != 200:
    print(f"Error: {response.status_code}")
    print(response.text)
else:
    # 2. Robust Parsing of RDB
    lines = [line for line in response.text.splitlines() if not line.startswith('#')]
    
    if len(lines) > 0:
        df = pd.read_csv(StringIO("\n".join(lines)), sep='\t')
        
        # Only drop if the dataframe isn't empty and contains the 15s/20s row
        if not df.empty:
            df = df.drop(0) 

            # 3. Clean Coordinates
            df['dec_lat_va'] = pd.to_numeric(df['dec_lat_va'], errors='coerce')
            df['dec_long_va'] = pd.to_numeric(df['dec_long_va'], errors='coerce')
            df = df.dropna(subset=['dec_lat_va', 'dec_long_va'])

            # Add mock RSLR data for the Viridis scale
            df['rslr_trend'] = np.random.uniform(2.8, 5.5, size=len(df))

            # 4. Plotting
            fig = plt.figure(figsize=(12, 9))
            ax = plt.axes(projection=ccrs.PlateCarree())
            ax.set_extent([-77.5, -75.0, 36.5, 40.0])

            ax.add_feature(cfeature.LAND, facecolor='#f0f0f0')
            ax.add_feature(cfeature.OCEAN, facecolor='#e0f7fa')
            ax.add_feature(cfeature.COASTLINE, linewidth=1)
            ax.add_feature(cfeature.RIVERS, edgecolor='blue', alpha=0.3)

            sc = ax.scatter(df['dec_long_va'], df['dec_lat_va'], 
                            c=df['rslr_trend'], cmap='viridis', 
                            s=40, transform=ccrs.PlateCarree(), 
                            edgecolor='black', alpha=0.8, zorder=5)

            plt.colorbar(sc, label='Relative Sea Level Rise Rate (mm/yr)', shrink=0.6)
            plt.title('USGS Monitoring Sites: Chesapeake Bay Region', fontsize=14)
            plt.show()
        else:
            print("The API returned a valid header but no station data for this box.")
    else:
        print("No data received from USGS.")
        
```

```python
import requests
import pandas as pd
from io import StringIO

usgs_url = "https://waterservices.usgs.gov/nwis/site/"
params = {
    'format': 'rdb',
    'bBox': '-77.5,36.0,-75.0,40.0', # Slightly wider box
    'seriesCatalogOutput': 'true',
    'siteStatus': 'active'
    # Removed parameterCode to catch everything
}

print("Requesting data from USGS...")
response = requests.get(usgs_url, params=params)

if response.status_code != 200:
    print(f"API Error: {response.status_code}")
else:
    lines = [l for l in response.text.splitlines() if not l.startswith('#')]
    print(f"Received {len(lines)} lines of data (including headers).")

    if len(lines) > 1:
        df = pd.read_csv(StringIO("\n".join(lines)), sep='\t')
        
        # Check for that USGS header row
        if not df.empty and str(df.iloc[0, 0]).endswith('s'):
            df = df.drop(0).reset_index(drop=True)
            
        print(f"Successfully loaded {len(df)} rows into DataFrame.")
        
        # Let's see what parameters are actually available in this area
        if 'parm_cd' in df.columns:
            print("\nAvailable Parameters in this region:")
            print(df['parm_cd'].value_counts().head(10))
        
        # Filter for Gage Height (00065) manually after loading
        df_tidal = df[df['parm_cd'] == '00065'].copy()
        print(f"\nFound {len(df_tidal)} stations with Gage Height (00065).")
    else:
        print("The API response contained no data rows.")
```

```python
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from io import StringIO
from datetime import datetime

# 1. API Call
usgs_url = "https://waterservices.usgs.gov/nwis/site/"
params = {
    'format': 'rdb',
    'bBox': '-77.5,36.0,-75.0,40.0',
    'seriesCatalogOutput': 'true',
    'siteStatus': 'active',
    'parameterCode': '00065' 
}

response = requests.get(usgs_url, params=params)

# 2. Parsing with Data Type Management
lines = [l for l in response.text.splitlines() if not l.startswith('#')]
if len(lines) > 1:
    # low_memory=False handles the DtypeWarning you saw
    df = pd.read_csv(StringIO("\n".join(lines)), sep='\t', low_memory=False)
    
    # Remove the USGS '5s' row
    if not df.empty and str(df.iloc[0, 0]).endswith('s'):
        df = df.drop(0).reset_index(drop=True)

    # Convert essential columns to numbers/dates
    df['dec_lat_va'] = pd.to_numeric(df['dec_lat_va'], errors='coerce')
    df['dec_long_va'] = pd.to_numeric(df['dec_long_va'], errors='coerce')
    df['begin_date'] = pd.to_datetime(df['begin_date'], errors='coerce')
    df['end_date'] = pd.to_datetime(df['end_date'], errors='coerce')
    
    df = df.dropna(subset=['dec_lat_va', 'dec_long_va', 'begin_date', 'end_date'])

    # 3. Logic: Calculate 'Years of Record'
    # This represents how much history we have to see a sea-level trend
    df['years_of_record'] = (df['end_date'] - df['begin_date']).dt.days / 365.25

    # 4. Map Setup
    fig = plt.figure(figsize=(12, 10))
    ax = plt.axes(projection=ccrs.PlateCarree())
    ax.set_extent([-77.8, -74.8, 36.5, 39.8])

    # Geography
    ax.add_feature(cfeature.COASTLINE, linewidth=1)
    ax.add_feature(cfeature.STATES, linestyle=':', alpha=0.5)
    ax.add_feature(cfeature.OCEAN, facecolor='#e3f2fd')
    ax.add_feature(cfeature.LAND, facecolor='#f5f5f5')

    # 5. Plot with Viridis
    # We cap the color scale at 30 years since that's the RSLR standard
    sc = ax.scatter(df['dec_long_va'], df['dec_lat_va'], 
                    c=df['years_of_record'], cmap='viridis', 
                    s=df['years_of_record']*4 + 10, 
                    vmin=0, vmax=30,
                    transform=ccrs.PlateCarree(), 
                    edgecolor='black', alpha=0.8, zorder=5)

    plt.colorbar(sc, label='Longevity of Record (Years)', shrink=0.5)
    plt.title('Non-NOAA (USGS) Water Level Monitoring\nChesapeake Bay Region', fontsize=14)
    
    plt.show()

    # Find the "Grandfather" station
    top_site = df.sort_values('years_of_record', ascending=False).iloc[0]
    print(f"\nLongest record found: {top_site['station_nm']}")
    print(f"Record Length: {top_site['years_of_record']:.1f} years (Started: {top_site['begin_date'].year})")
```
