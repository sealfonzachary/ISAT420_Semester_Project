```python
# --- 1. AREA SETUP (minx, miny, maxx, maxy) & DATA RETRIEVAL ---
import warnings
# Suppress all FutureWarnings (they're super annyoing and clutter the output)
warnings.simplefilter(action='ignore', category=FutureWarning)
import pandas as pd
import pygeohydro as gh
import geopandas as gpd
from shapely.geometry import box
from shapely.geometry import Point
import matplotlib.pyplot as plt
import numpy as np
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# Create a bounding box for our data
bbox = [-77.5, 36.5, -75.0, 40.0]

# Convert the bounding box into a GeoDataFrame, defining the CRS as 4326 (standard Latitude/Longitude)
geometry = gpd.GeoDataFrame({'geometry': [box(*bbox)]}, crs="EPSG:4326")

# Fetch NLCD data using nlcd_bygeom, providing the resolution (in meters) and the years in a dictionary
# The result is a dictionary, so we access it with the key 0
print("Fetching NLCD data from USGS...")
nlcd_dict = gh.nlcd_bygeom(geometry, resolution=30, years={'cover': [2001, 2021]})
nlcd = nlcd_dict[0] # The 0 index corresponds to our geometry
print("...Fetching complete!")
```

```python
# --- 2. SEPARATE YEARS, CLASSES, AND MAKE A FILTER ---

# The returned object will have 'cover_2001' and 'cover_2021' variables
ds2001 = nlcd['cover_2001']
ds2021 = nlcd['cover_2021']

# Define classes (41, 42, and 43 are forest classes, 90 and 95 are wetland classes)
was_forest = ds2001.isin([41, 42, 43])
is_developed = ds2021.isin([90, 95])

# Our filter asks: which pixels WERE a forest in 2001 and are NOW a wetland in 2021?
ghost_forest_mask = was_forest & is_wetland
```

```python
# --- 3. APPLY FILTER AND PLOT ---
# Create the transition map
transition_map = ghost_forest_mask.where(ghost_forest_mask == True, np.nan).astype(float)

# Setup with a Cartopy projection
fig = plt.figure(figsize=(10, 8))
ax = plt.axes(projection=ccrs.PlateCarree()) 

# Adding coastline and geographic features
ax.add_feature(cfeature.COASTLINE, linewidth=1, edgecolor='black')
ax.add_feature(cfeature.STATES, linestyle=':', edgecolor='gray')

# Plotting the data
# We use transform=ccrs.PlateCarree() because the data is in lat/lon degrees
transition_map.plot(
    ax=ax, 
    cmap='Reds', 
    add_colorbar=False, 
    transform=ccrs.PlateCarree()
)

# Use the bbox defined earlier in the code
ax.set_extent([-77.5, -75.0, 36.5, 40.0], crs=ccrs.PlateCarree())

plt.title('Potential Ghost Forest Transition (2001-2021)')
plt.show()

# Statistical count
count = np.nansum(transition_map.values)
print(f"Number of 30m pixels identified as ghost forest transition: {count}")
```

```python
# --- 4. CREATE A DENSITY MAP TO SEE THE TRANSITION MORE CLEARLY ---
# Stack the dimensions 'x' and 'y' into one dimension called 'pixel'
# Drop all the NaNs (so we only keep the 'True' / 1.0 pixels)
valid_pixels = transition_map.stack(pixel=('x', 'y')).dropna('pixel')

# Extract the X and Y coordinates from the valid pixels
lons = valid_pixels.x.values
lats = valid_pixels.y.values

# Plot the density map
plt.figure(figsize=(10, 8))
ax = plt.axes(projection=ccrs.PlateCarree())
ax.add_feature(cfeature.COASTLINE)

# Now these arrays are guaranteed to be the same length
hb = ax.hexbin(lons, lats, gridsize=100, cmap='inferno', mincnt=1, transform=ccrs.PlateCarree())

plt.colorbar(hb, label='Number of Transition Pixels')
plt.title('Ghost Forest Transition Density')
plt.show()
```

```python
# --- 1. LOAD DATA ---
dfrslr = r"../TidalGaugeTest/RSLR_Trend_Summary.csv"
dfsalinity = r"../TidalSalinityTest/Salinity_Trend_Summary.csv"
rslr_df = pd.read_csv(dfrslr)
salinity_df = pd.read_csv(dfsalinity)

# --- 2. CONVERT TO GEOPANDAS ---
def to_gdf(df, lon_col='Lon', lat_col='Lat'):
    geometry = [Point(xy) for xy in zip(df[lon_col], df[lat_col])]
    gdf = gpd.GeoDataFrame(df, geometry=geometry, crs="EPSG:4326")
    # Project to UTM 18N (meters) for accurate 25km distance calculation
    return gdf.to_crs("EPSG:26918")

rslr_gdf = to_gdf(rslr_df)
salinity_gdf = to_gdf(salinity_df)

# --- 3. CONVERT NLCD PIXELS TO POINTS ---
# Assuming 'valid_pixels' from your previous code:
pixel_gdf = gpd.GeoDataFrame(
    geometry=[Point(xy) for xy in zip(lons, lats)], # lons/lats from NLCD script
    crs="EPSG:4326"
).to_crs("EPSG:26918")

# --- 4. CREATE 25KM BUFFERS ---
# Buffer is in meters, so 25000 = 25km
rslr_gdf['geometry_buffer'] = rslr_gdf.geometry.buffer(50000)

# --- 5. SPATIAL COUNTS ---
results = []

for idx, station in rslr_gdf.iterrows():
    circle = station['geometry_buffer']
    
    # Count Salinity stations inside this circle
    sal_inside = salinity_gdf[salinity_gdf.geometry.within(circle)]
    avg_sal_rate = sal_inside['Rate (PPT/yr)'].mean() if not sal_inside.empty else 0
    
    # Count Ghost Forest pixels inside this circle
    pixel_count = pixel_gdf[pixel_gdf.geometry.within(circle)].shape[0]
    
    results.append({
        'RSLR_Station': station['Station ID'],
        'RSLR_Rate': station['Rate (mm/yr)'],
        'Avg_Salinity_Trend': avg_sal_rate,
        'Ghost_Forest_Pixels': pixel_count,
        'Salinity_Station_Count': len(sal_inside)
    })

# --- 6. FINAL DATASET ---
comparison_df = pd.DataFrame(results)
print(comparison_df)
```

```python
# Calculate correlation matrix
correlation = comparison_df[['RSLR_Rate', 'Avg_Salinity_Trend', 'Ghost_Forest_Pixels']].corr()

print("--- Correlation with Ghost Forest Pixels ---")
print(correlation['Ghost_Forest_Pixels'])
```

```python
import numpy as np
import pandas as pd
import geopandas as gpd
import statsmodels.api as sm
from scipy.spatial import cKDTree

# 1. Prepare and Project the Data
# Using EPSG:3857 (Web Mercator) ensures units are in meters
target_crs = "EPSG:3857"
rslr_proj = rslr_gdf.to_crs(target_crs)
slope_points_proj = gpd.read_file("./usgs_slope_data/POINTS_CBslope/POINTS_CBslope.shp").to_crs(target_crs)

# 2. Extract Coordinates for the Search
# We pull the XY coordinates into a format the "Search Tree" can read
slope_coords = np.array(list(zip(slope_points_proj.geometry.x, slope_points_proj.geometry.y)))
station_coords = np.array(list(zip(rslr_proj.geometry.x, rslr_proj.geometry.y)))

# 3. Build the Search Tree and Find Neighbors
# This finds the 10 closest USGS slope points for every RSLR station
tree = cKDTree(slope_coords)
distances, indices = tree.query(station_coords, k=10)

# 4. Calculate Mean Slope from those 10 neighbors
mean_slopes = []
for idx_list in indices:
    # Grab the 'Avg_Slope' values for the 10 closest points found in the tree
    neighbor_slopes = slope_points_proj.iloc[idx_list]['Avg_Slope']
    mean_slopes.append(neighbor_slopes.mean())

# 5. Add to your main dataframe
comparison_df['Mean_Slope'] = mean_slopes

print("--- Success! Slope values joined by proximity ---")
print(comparison_df[['RSLR_Station', 'Ghost_Forest_Pixels', 'Mean_Slope']].head())

# 6. RUN THE FINAL MULTIVARIATE REGRESSION
print("\n--- Running Final Regression (RSLR + Salinity + Slope) ---")
X = comparison_df[['RSLR_Rate', 'Avg_Salinity_Trend', 'Mean_Slope']]
X = sm.add_constant(X)
y = comparison_df['Ghost_Forest_Pixels']

final_model = sm.OLS(y, X).fit()
print(final_model.summary())
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# --- 1. RADIUS SEARCH (The Scale Fix) ---
# Find EVERY USGS point within 25,000 meters of each station
indices = tree.query_ball_point(station_coords, r=25000)

better_slopes = []
for idx_list in indices:
    if len(idx_list) > 0:
        # Average ALL USGS points found in the 25km circle
        better_slopes.append(slope_points_proj.iloc[idx_list]['Avg_Slope'].mean())
    else:
        # If no points found (unlikely), use the global average
        better_slopes.append(np.nan)

comparison_df['Mean_Slope_25km'] = better_slopes
comparison_df['Mean_Slope_25km'] = comparison_df['Mean_Slope_25km'].fillna(comparison_df['Mean_Slope_25km'].mean())

# --- 2. RE-SPLIT DATA (The KeyError Fix) ---
# Now that comparison_df has 'Mean_Slope_25km', we re-create the subsets
mainstem_df = comparison_df[comparison_df['Area'] == 'Mainstem']
tributary_df = comparison_df[comparison_df['Area'] == 'Tributary']

# --- 3. CALCULATE CORRELATIONS ---
print("\n--- Correlation: Mainstem (25km Landscape Slope) ---")
cols = ['RSLR_Rate', 'Avg_Salinity_Trend', 'Mean_Slope_25km', 'Ghost_Forest_Pixels']
print(mainstem_df[cols].corr()['Ghost_Forest_Pixels'])

print("\n--- Correlation: Tributaries (25km Landscape Slope) ---")
print(tributary_df[cols].corr()['Ghost_Forest_Pixels'])

# --- 4. VISUALIZING SLOPE BY AREA ---
plt.figure(figsize=(8, 6))
sns.boxplot(x='Area', y='Mean_Slope_25km', data=comparison_df, hue='Area', palette='viridis', legend=False)
plt.title('Landscape Steepness (25km Radius): Mainstem vs. Tributaries')
plt.ylabel('Average Slope (%)')
plt.show()
```

```python
from scipy import stats

# 1. T-test for Ghost Forest Pixels (Are Tributaries actually ghostier?)
t_stat_forest, p_val_forest = stats.ttest_ind(
    mainstem_df['Ghost_Forest_Pixels'], 
    tributary_df['Ghost_Forest_Pixels'], 
    equal_var=False
)

# 2. T-test for Slope (Are Tributaries actually flatter?)
t_stat_slope, p_val_slope = stats.ttest_ind(
    mainstem_df['Mean_Slope_25km'], 
    tributary_df['Mean_Slope_25km'], 
    equal_var=False
)

print(f"Ghost Forest P-Value: {p_val_forest:.4f}")
print(f"Mean Slope P-Value: {p_val_slope:.4f}")
```

```python
import folium
from branca.colormap import LinearColormap

# 1. Prepare the Map GeoDataFrame and ensure IDs are strings
map_gdf = rslr_gdf.copy().to_crs("EPSG:4326")
map_gdf['Station ID'] = map_gdf['Station ID'].astype(str)

# 2. Ensure comparison_df IDs are also strings
comparison_df['RSLR_Station'] = comparison_df['RSLR_Station'].astype(str)

# 3. Merge (This should now work without the ValueError)
map_data = map_gdf.merge(
    comparison_df[['RSLR_Station', 'Ghost_Forest_Pixels', 'Mean_Slope_25km', 'Avg_Salinity_Trend', 'Area']], 
    left_on='Station ID', 
    right_on='RSLR_Station'
)

# 4. Initialize Map
m = folium.Map(location=[38.5, -76.3], zoom_start=8, tiles='CartoDB positron')

# 5. Set up the Slope Gradient
colormap = LinearColormap(
    colors=['red', 'yellow', 'green'], 
    vmin=map_data['Mean_Slope_25km'].min(), 
    vmax=map_data['Mean_Slope_25km'].max()
)
colormap.caption = 'Landscape Slope (%) - Red is flatter (The Ramp)'

# 6. Add Circles to Map
for idx, row in map_data.iterrows():
    # Folium expects [Lat, Lon]
    lat, lon = row.geometry.y, row.geometry.x
    color = colormap(row['Mean_Slope_25km'])
    
    folium.Circle(
        location=[lat, lon],
        radius=25000, 
        color=color,
        fill=True,
        fill_opacity=0.4,
        popup=(f"<b>Station:</b> {row['RSLR_Station']}<br>"
               f"<b>Area:</b> {row['Area']}<br>"
               f"<b>Ghost Forest Pixels:</b> {row['Ghost_Forest_Pixels']:.0f}<br>"
               f"<b>Mean Slope (25km):</b> {row['Mean_Slope_25km']:.2f}%<br>"
               f"<b>Salinity Trend:</b> {row['Avg_Salinity_Trend']:.4f}")
    ).add_to(m)

m.add_child(colormap)
m.save('Chesapeake_Ghost_Forest_Final_Analysis.html')
print("Victory! Map saved to 'Chesapeake_Ghost_Forest_Final_Analysis.html'")
m
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from matplotlib.patches import Circle
from matplotlib.lines import Line2D
import matplotlib.cm as cm

# 1. Setup the map projection
fig = plt.figure(figsize=(14, 14))
# Use PlateCarree for lat/lon, but Mercator is often better for coastal views
ax = plt.axes(projection=ccrs.Mercator())

# 2. Add high-resolution geographic context
ax.add_feature(cfeature.COASTLINE.with_scale('10m'), linewidth=1)
ax.add_feature(cfeature.STATES.with_scale('10m'), edgecolor='gray', alpha=0.5)
ax.add_feature(cfeature.LAND, facecolor='#fdfdfd')
ax.add_feature(cfeature.OCEAN, facecolor='#f0f8ff') # Add some blue to the water

# Set extent to focus on the study areas
ax.set_extent([-77.0, -75.0, 36.8, 39.5], crs=ccrs.PlateCarree())

# --- 3. NEW LAYER: The Slope Gradient (Coolwarm) ---
# We use tripcolor to create a smooth gradient from the raw USGS points
# slope_points_proj must contain: geometry (x, y in EPSG:3857) and 'Avg_Slope'
slope_x = slope_points_proj.geometry.x
slope_y = slope_points_proj.geometry.y
slope_val = slope_points_proj['Avg_Slope']

# Use diverging color map: Red (Low Ramp) -> White -> Blue (High Wall)
sc = ax.tripcolor(slope_x, slope_y, slope_val, cmap='coolwarm', 
                  shading='gouraud', alpha=0.6, transform=ccrs.Mercator())

# Add Colorbar for Slope
cbar = plt.colorbar(sc, ax=ax, orientation='vertical', shrink=0.7, pad=0.02)
cbar.set_label('Geomorphic Slope (%)', rotation=270, labelpad=15, fontsize=12)

# --- 4. NEW LAYER: Ghost Forest Density (Purple) ---
# We use purple with high transparency to show density clusters
# lons, lats are from your NLCD analysis
ax.scatter(lons, lats, s=1, color='purple', alpha=0.05, 
           transform=ccrs.PlateCarree(), zorder=3, label='Ghost Forest Pixels')

# 5. Loop through RSLR stations
for idx, row in comparison_df.iterrows():
    # Get original lat/lon from the RSLR dataframe
    lon, lat = rslr_df.iloc[idx]['Lon'], rslr_df.iloc[idx]['Lat']
    
    # --- NEW: Pink Buffers for High Contrast ---
    # Draw the 25km buffer circle in bold PINK
    circle_patch = plt.Circle((lon, lat), 0.225, color='#FF1493', fill=False, 
                              linewidth=2.5, alpha=0.7, transform=ccrs.PlateCarree(), zorder=4)
    ax.add_patch(circle_patch)
    
    # Plot Salinity Stations (Green Triangles)
    circle_geom = rslr_gdf.iloc[idx]['geometry_buffer']
    sal_inside = salinity_gdf[salinity_gdf.geometry.within(circle_geom)]
    sal_inside_latlon = sal_inside.to_crs("EPSG:4326")
    
    ax.scatter(sal_inside_latlon.geometry.x, sal_inside_latlon.geometry.y, 
               marker='^', color='green', s=50, edgecolors='white', linewidth=0.5,
               transform=ccrs.PlateCarree(), zorder=5)
    
    # Plot the RSLR tidal station point (the center)
    ax.plot(lon, lat, 'ko', markersize=8, transform=ccrs.PlateCarree(), zorder=6)
    
    # Add labeled station ID, also with pink contrast if you want
    ax.text(lon + 0.04, lat + 0.04, str(idx), fontsize=11, weight='bold',
            transform=ccrs.PlateCarree(), bbox=dict(facecolor='white', alpha=0.8, edgecolor='#FF1493', boxstyle='round,pad=0.3'), zorder=7)

# 6. Create the Expert Legend
legend_elements = [
    Line2D([0], [0], marker='o', color='w', label='RSLR Station (Anchor)', markerfacecolor='black', markersize=10),
    Line2D([0], [0], marker='^', color='w', label='Salinity Station', markerfacecolor='green', markersize=10),
    Line2D([0], [0], marker='.', color='purple', alpha=0.4, label='Ghost Forest Pixel Density', markersize=15, linewidth=0),
    Line2D([0], [0], color='#FF1493', lw=2.5, alpha=0.7, label='25km Geomorphic Buffer')
]
ax.legend(handles=legend_elements, loc='upper right', fontsize=11, framealpha=0.9)

plt.title("Geomorphic Context of Forest Loss: Chesapeake Bay\nSlope (Ramp vs. Wall) and Salinity Dynamics", fontsize=18, weight='bold')
plt.show()

# --- 7. REFINED TABLE OUTPUT ---
print(f"{'No.':<4} | {'Station ID':<12} | {'Pixels':<8} | {'Slope (%)':<10} | {'Salinity PPT/Yr'}")
print("-" * 65)
for idx, row in comparison_df.iterrows():
    print(f"{idx:<4} | {row['RSLR_Station']:<12} | {int(row['Ghost_Forest_Pixels']):<8} | {row['Mean_Slope_25km']:<10.2f} | {row['Avg_Salinity_Trend']:.4f}")
```
